```python
import pandas as pd
import os
import glob
```

```python
base_dir = '/Users/cherylxiang/Documents/GitHub/demultiplexing-methods'
files = glob.glob(f'{base_dir}/results/demultiplex/*/classifications.csv')

all_classifications = []
for f in files:
    dataset_id = f.split('/')[-2]  # extract dataset name from path
    df = pd.read_csv(f)
    df['dataset'] = dataset_id
    df['method'] = 'demultiplex'
    all_classifications.append(df)

demultiplex_classifications = pd.concat(all_classifications, ignore_index=True)
print(demultiplex_classifications['dataset'].unique())
demultiplex_classifications.head()
```
