```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import glob
from pdf2image import convert_from_path
```

```python
pdf_files = sorted(glob.glob('analysis/parameter_analysis/figures/*.pdf'))

for pdf in pdf_files:
    images = convert_from_path(pdf)
    for img in images:
        plt.figure(figsize=(10, 8))
        plt.imshow(img)
        plt.axis('off')
        plt.title(pdf.split('/')[-1])
        plt.tight_layout()
        plt.show()
```
